```python
import numpy as np

# Example input: three extracted face features
x = np.array([[0.8, 0.2, 0.6]])

# Weights and biases: input layer (3) -> hidden layer (4)
W1 = np.array([
    [ 0.2, -0.4,  0.1,  0.3],
    [-0.3,  0.5,  0.2, -0.1],
    [ 0.4,  0.1, -0.2,  0.2],
])
b1 = np.array([0.1, -0.1, 0.05, 0.0])

# Weights and biases: hidden layer (4) -> output layer (3 identities)
W2 = np.array([
    [ 0.3, -0.2,  0.1],
    [-0.1,  0.4,  0.2],
    [ 0.2,  0.1, -0.3],
    [ 0.5, -0.2,  0.2],
])
b2 = np.array([0.0, 0.1, -0.1])


def sigmoid(z):
    return 1 / (1 + np.exp(-z))


def softmax(z):
    # Subtract the maximum for numerical stability
    exp_z = np.exp(z - np.max(z, axis=1, keepdims=True))
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)


# Hidden layer forward propagation
z_hidden = x @ W1 + b1
hidden_output = sigmoid(z_hidden)

# Output layer forward propagation
z_output = hidden_output @ W2 + b2
output_probabilities = softmax(z_output)

print("Hidden layer output:")
print(hidden_output)

print("\nOutput probabilities:")
print(output_probabilities)

predicted_identity = np.argmax(output_probabilities, axis=1)[0]
print(f"\nPredicted identity: Person {predicted_identity + 1}")
```
