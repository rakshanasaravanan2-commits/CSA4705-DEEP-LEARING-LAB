```python
import numpy as np
from sklearn.neural_network import MLPClassifier

# XOR input patterns
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# XOR output
y = np.array([0, 1, 1, 0])

# Feedforward neural network with one hidden layer
model = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation="relu",
    solver="lbfgs",
    max_iter=1000,
    random_state=42
)

# Train model
model.fit(X, y)

# Predict XOR outputs
predictions = model.predict(X)

print("Input\tExpected\tPredicted")
for inputs, expected, predicted in zip(X, y, predictions):
    print(f"{inputs}\t{expected}\t\t{predicted}")
```
