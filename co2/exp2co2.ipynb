```python
# Polynomial Regression: Bias–Variance Trade-off for House Price Prediction

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Sample house data: area (sq. ft.) vs price
np.random.seed(42)
area = np.linspace(500, 5000, 150).reshape(-1, 1)
price = 50 * area.flatten() + 0.01 * (area.flatten() ** 2) + np.random.normal(0, 50000, 150)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    area, price, test_size=0.30, random_state=42
)

degrees = [1, 2, 3, 5, 10]
results = []

for degree in degrees:
    polynomial = PolynomialFeatures(degree=degree)
    X_train_poly = polynomial.fit_transform(X_train)
    X_test_poly = polynomial.transform(X_test)

    model = LinearRegression()
    model.fit(X_train_poly, y_train)

    train_prediction = model.predict(X_train_poly)
    test_prediction = model.predict(X_test_poly)

    train_error = mean_squared_error(y_train, train_prediction)
    test_error = mean_squared_error(y_test, test_prediction)

    results.append((degree, train_error, test_error))

# Display comparison
print("Degree | Training MSE | Testing MSE | Evaluation")
print("-" * 58)

best_degree = min(results, key=lambda row: row[2])[0]

for degree, train_error, test_error in results:
    if degree == 1:
        evaluation = "High bias: likely underfitting"
    elif degree == best_degree:
        evaluation = "Best balance: selected model"
    else:
        evaluation = "High variance: possible overfitting"

    print(f"{degree:>6} | {train_error:>12.2f} | {test_error:>11.2f} | {evaluation}")

print(f"\nRecommended polynomial degree: {best_degree}")
print("Choose the model with the lowest testing MSE because it generalizes best.")
```
