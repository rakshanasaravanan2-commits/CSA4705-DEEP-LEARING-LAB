```python
import numpy as np
import matplotlib.pyplot as plt


def sigmoid(x):
    # Clip to avoid overflow when computing exp(-x)
    x = np.clip(x, -500, 500)
    return 1 / (1 + np.exp(-x))


def tanh(x):
    return np.tanh(x)


def relu(x):
    return np.maximum(0, x)


# Input values to compare
x = np.linspace(-6, 6, 500)

activations = {
    "Sigmoid": sigmoid(x),
    "Tanh": tanh(x),
    "ReLU": relu(x),
}

# Plot each activation function
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, y) in zip(axes, activations.items()):
    ax.plot(x, y, linewidth=2)
    ax.axhline(0, color="gray", linewidth=0.8)
    ax.axvline(0, color="gray", linewidth=0.8)
    ax.set_title(name)
    ax.set_xlabel("Input")
    ax.set_ylabel("Output")
    ax.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

# Print a short comparison
print("Activation function comparison")
print("- Sigmoid: output is between 0 and 1; useful for probabilities,")
print("  but its gradient becomes small for large positive or negative inputs.")
print("- Tanh: output is between -1 and 1 and centered around zero;")
print("  it also has small gradients for inputs with large magnitude.")
print("- ReLU: output is 0 for negative inputs and equals the input for")
print("  positive inputs; it is simple and avoids saturation on the positive side,")
print("  but can produce zero output and gradient for negative inputs.")
```
