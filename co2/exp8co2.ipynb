```python
import numpy as np

# Input: [Income, Credit Score]
X = np.array([
    [25000, 550],
    [30000, 600],
    [45000, 650],
    [50000, 700],
    [60000, 720],
    [70000, 780],
    [35000, 580],
    [80000, 800]
])

# Output: 0 = Ineligible, 1 = Eligible
y = np.array([0, 0, 1, 1, 1, 1, 0, 1])

# Normalize input values
X = (X - X.mean(axis=0)) / X.std(axis=0)

# Initialize weights and bias
weights = np.zeros(2)
bias = 0
learning_rate = 0.1
epochs = 20

def step_function(value):
    return 1 if value >= 0 else 0

# Train single-layer perceptron
for epoch in range(epochs):
    for inputs, target in zip(X, y):
        prediction = step_function(np.dot(inputs, weights) + bias)
        error = target - prediction

        weights += learning_rate * error * inputs
        bias += learning_rate * error

print("Trained Weights:", weights)
print("Bias:", bias)

# Test applicant: Income = 55,000, Credit Score = 710
new_applicant = np.array([[55000, 710]])
new_applicant = (new_applicant - np.array([46875, 672.5])) / np.array([18150, 85])

prediction = step_function(np.dot(new_applicant[0], weights) + bias)

if prediction == 1:
    print("Loan Applicant: Eligible")
else:
    print("Loan Applicant: Ineligible")
```
