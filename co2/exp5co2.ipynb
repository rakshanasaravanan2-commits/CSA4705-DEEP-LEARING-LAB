```python
# Floating-Point Precision Error Demonstration

import numpy as np
from decimal import Decimal, getcontext

# Set high precision for Decimal calculations
getcontext().prec = 50

print("1. Precision error: 0.1 + 0.2")
print("float64 :", 0.1 + 0.2)
print("Expected:", 0.3)
print("Equal?  :", (0.1 + 0.2) == 0.3)

print("\n2. Comparing float32 and float64")
value32 = np.float32(1.0 / 3.0)
value64 = np.float64(1.0 / 3.0)

print("float32:", value32)
print("float64:", value64)
print("Difference:", abs(value64 - value32))

print("\n3. Large number + small number")
large32 = np.float32(1e10)
small32 = np.float32(1.0)

large64 = np.float64(1e10)
small64 = np.float64(1.0)

print("float32: 1e10 + 1 =", large32 + small32)
print("float64: 1e10 + 1 =", large64 + small64)

print("\n4. Repeated addition error")
sum32 = np.float32(0.0)
sum64 = np.float64(0.0)

for _ in range(1_000_000):
    sum32 += np.float32(0.1)
    sum64 += np.float64(0.1)

expected = 100000.0

print("Expected value:", expected)
print("float32 sum :", sum32, "| Error:", abs(expected - sum32))
print("float64 sum :", sum64, "| Error:", abs(expected - sum64))

print("\n5. High-precision Decimal calculation")
decimal_sum = Decimal("0.1") + Decimal("0.2")
print("Decimal 0.1 + 0.2 =", decimal_sum)
```
