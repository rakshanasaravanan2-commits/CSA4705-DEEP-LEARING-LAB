```python
# Overfitting analysis: student exam performance prediction

from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error

# Create sample student-performance data
X, y = make_regression(n_samples=200, n_features=5, noise=25, random_state=42)

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)

# Complex model: likely to overfit
overfit_model = make_pipeline(
    PolynomialFeatures(degree=5),
    LinearRegression()
)
overfit_model.fit(X_train, y_train)

train_error = mean_squared_error(y_train, overfit_model.predict(X_train))
test_error = mean_squared_error(y_test, overfit_model.predict(X_test))

print("Overfitting model")
print("Training MSE:", round(train_error, 2))
print("Testing MSE :", round(test_error, 2))

if test_error > train_error * 1.5:
    print("\nResult: Overfitting detected.")
    print("The model performs very well on training data but poorly on unseen test data.")

# Improved model: lower complexity + Ridge regularization
improved_model = make_pipeline(
    PolynomialFeatures(degree=2),
    Ridge(alpha=10)
)
improved_model.fit(X_train, y_train)

improved_train_error = mean_squared_error(
    y_train, improved_model.predict(X_train)
)
improved_test_error = mean_squared_error(
    y_test, improved_model.predict(X_test)
)

print("\nImproved model")
print("Training MSE:", round(improved_train_error, 2))
print("Testing MSE :", round(improved_test_error, 2))

print("\nImprovements:")
print("- Reduce model complexity")
print("- Apply regularization (Ridge/Lasso)")
print("- Use cross-validation")
print("- Collect more student data")
print("- Remove irrelevant features")
```
