```python
import numpy as np

# Create a 3x3 image matrix
A = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

# Flatten the image matrix
B = A.flatten()

# Calculate total pixel intensity
total = np.sum(A)

# Display results
print("Original Image Matrix:")
print(A)

print("Flattened Image Vector:")
print(B)

print("Total Pixel Intensity:")
print(total)
```
