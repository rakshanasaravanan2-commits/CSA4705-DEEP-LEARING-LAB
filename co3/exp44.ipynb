```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, BatchNormalization, ReLU

# Load MNIST dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Preprocessing
X_train = X_train.reshape(-1, 28, 28, 1).astype('float32') / 255.0
X_test = X_test.reshape(-1, 28, 28, 1).astype('float32') / 255.0

# Model without Batch Normalization
model1 = Sequential([
    tf.keras.Input(shape=(28, 28, 1)),
    Conv2D(32, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

# Model with Batch Normalization
model2 = Sequential([
    tf.keras.Input(shape=(28, 28, 1)),
    Conv2D(32, (3, 3)),
    BatchNormalization(),
    ReLU(),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3)),
    BatchNormalization(),
    ReLU(),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128),
    BatchNormalization(),
    ReLU(),
    Dense(10, activation='softmax')
])

# Compile models
for model in [model1, model2]:
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

# Train without Batch Normalization
print("Without Batch Normalization")
model1.fit(X_train, y_train, epochs=5, batch_size=64, verbose=1)
loss1, acc1 = model1.evaluate(X_test, y_test, verbose=0)

# Train with Batch Normalization
print("\nWith Batch Normalization")
model2.fit(X_train, y_train, epochs=5, batch_size=64, verbose=1)
loss2, acc2 = model2.evaluate(X_test, y_test, verbose=0)

# Compare results
print("\nPerformance Comparison")
print("Without BN - Accuracy:", acc1 * 100, "%")
print("With BN - Accuracy:", acc2 * 100, "%")
print("Without BN - Loss:", loss1)
print("With BN - Loss:", loss2)
```
