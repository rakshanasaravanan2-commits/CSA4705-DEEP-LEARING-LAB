```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.layers import RandomRotation, RandomFlip, RandomZoom

# Load MNIST dataset (sample image dataset)
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Preprocessing
X_train = X_train.reshape(-1, 28, 28, 1).astype('float32') / 255.0
X_test = X_test.reshape(-1, 28, 28, 1).astype('float32') / 255.0

# Image Augmentation
augmentation = Sequential([
    RandomRotation(0.1),
    RandomFlip("horizontal"),
    RandomZoom(0.1)
])

# Create CNN model
def create_model(use_augmentation=False):

    model = Sequential()

    model.add(tf.keras.Input(shape=(28, 28, 1)))

    if use_augmentation:
        model.add(augmentation)

    model.add(Conv2D(32, (3, 3), activation='relu'))
    model.add(MaxPooling2D((2, 2)))
    model.add(Conv2D(64, (3, 3), activation='relu'))
    model.add(MaxPooling2D((2, 2)))
    model.add(Flatten())
    model.add(Dense(128, activation='relu'))
    model.add(Dense(10, activation='softmax'))

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

# Without Augmentation
model1 = create_model()
model1.fit(X_train, y_train, epochs=3, batch_size=64, verbose=1)
loss1, acc1 = model1.evaluate(X_test, y_test, verbose=0)

# With Augmentation
model2 = create_model(use_augmentation=True)
model2.fit(X_train, y_train, epochs=3, batch_size=64, verbose=1)
loss2, acc2 = model2.evaluate(X_test, y_test, verbose=0)

# Compare Performance
print("Without Augmentation Accuracy:", acc1 * 100, "%")
print("With Augmentation Accuracy:", acc2 * 100, "%")

print("Without Augmentation Loss:", loss1)
print("With Augmentation Loss:", loss2)
```
