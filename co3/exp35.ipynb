```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input

(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

X_train = X_train / 255.0
X_test = X_test / 255.0

def create_model(drop):
    model = Sequential([
        Input(shape=(28, 28)),
        tf.keras.layers.Flatten(),
        Dense(256, activation='relu'),
        Dropout(0.5) if drop else Dense(256, activation='relu'),
        Dense(128, activation='relu'),
        Dense(10, activation='softmax')
    ])

    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

model1 = create_model(False)
model2 = create_model(True)

print("Without Dropout")
model1.fit(X_train, y_train, epochs=5, verbose=0)
print(model1.evaluate(X_test, y_test, verbose=0))

print("With Dropout")
model2.fit(X_train, y_train, epochs=5, verbose=0)
print(model2.evaluate(X_test, y_test, verbose=0))
```
