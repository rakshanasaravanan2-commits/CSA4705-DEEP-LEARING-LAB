```python
import numpy as np

# Transition matrix: columns represent the page being linked from.
# Each column's values sum to 1.
transition_matrix = np.array([
    [0.0, 0.5, 1.0],  # Links to Page A
    [0.5, 0.0, 0.0],  # Links to Page B
    [0.5, 0.5, 0.0]   # Links to Page C
])

eigenvalues, eigenvectors = np.linalg.eig(transition_matrix)

# Select the eigenvector for the eigenvalue closest to 1
index = np.argmin(np.abs(eigenvalues - 1))
importance = np.real(eigenvectors[:, index])

# Normalize scores so they sum to 1
importance = np.abs(importance)
importance /= importance.sum()

print("Transition matrix:")
print(transition_matrix)
print("\nEigenvalues:")
print(eigenvalues)
print("\nEigenvectors:")
print(eigenvectors)
print("\nPage importance scores:")
for page, score in zip(["Page A", "Page B", "Page C"], importance):
    print(f"{page}: {score:.4f}")
```

```python
import numpy as np

customers = {
    "Customer A": np.array([5, 2, 0, 1]),
    "Customer B": np.array([4, 1, 0, 1]),
    "Customer C": np.array([0, 1, 5, 2])
}

def cosine_similarity(vector_a, vector_b):
    dot_product = np.dot(vector_a, vector_b)
    magnitude_a = np.linalg.norm(vector_a)
    magnitude_b = np.linalg.norm(vector_b)

    if magnitude_a == 0 or magnitude_b == 0:
        return 0.0, dot_product

    similarity = dot_product / (magnitude_a * magnitude_b)
    return similarity, dot_product

customer_names = list(customers)

for i in range(len(customer_names)):
    for j in range(i + 1, len(customer_names)):
        name_a = customer_names[i]
        name_b = customer_names[j]

        similarity, dot_product = cosine_similarity(
            customers[name_a], customers[name_b]
        )

        print(f"{name_a} and {name_b}:")
        print(f"  Dot product: {dot_product}")
        print(f"  Cosine similarity: {similarity:.4f}")
```
