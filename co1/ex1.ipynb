```python
import numpy as np

# Current robot position and movement vector in 3D
position = np.array([1.0, 2.0, 3.0])
movement = np.array([0.5, -1.0, 2.0])

# Vector addition
added = position + movement

# Vector subtraction
difference = position - movement

# Scalar multiplication
scaled_movement = 2.0 * movement

# Calculate the new position using the scaled movement
new_position = position + scaled_movement

print("Addition:", added)
print("Subtraction:", difference)
print("Scaled movement:", scaled_movement)
print("New position:", new_position)
```
