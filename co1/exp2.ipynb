```python
import numpy as np

# A transformation matrix and a matrix of image coordinates
transform = np.array([
    [2, 0, 1],
    [0, 3, 1],
    [0, 0, 1]
])

coordinates = np.array([
    [1, 2, 1],
    [3, 4, 1],
    [5, 6, 1]
]).T  # Each coordinate is a column: [x, y, 1]

# Matrix multiplication applies the transformation to each coordinate
transformed = transform @ coordinates

# Transpose swaps rows and columns
transposed = transformed.T

print("Transformation matrix:\n", transform)
print("Original coordinates (columns):\n", coordinates)
print("Transformed coordinates (columns):\n", transformed)
print("Transformed coordinates (rows):\n", transposed)
```
