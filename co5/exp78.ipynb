```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

mean = [0, 0]
cov = [[1, 0.5], [0.5, 1]]

data = np.random.multivariate_normal(mean, cov, 1000)

print("Generated Data:")
print(data[:5])

print("Mean:")
print(np.mean(data, axis=0))

print("Covariance:")
print(np.cov(data.T))

plt.scatter(data[:, 0], data[:, 1])
plt.title("Gaussian Distribution")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
