```python
import numpy as np
import matplotlib.pyplot as plt

gradient = 1.0

values = []

for i in range(20):
    gradient = gradient * 0.5
    values.append(gradient)

print("Gradient Values:")
print(values)

plt.plot(values)
plt.title("Vanishing Gradient Problem")
plt.xlabel("Time Steps")
plt.ylabel("Gradient")
plt.show()
```
